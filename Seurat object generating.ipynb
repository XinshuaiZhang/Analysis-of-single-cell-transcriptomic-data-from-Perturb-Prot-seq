## normal 1样本

### 读取sgRNA数据

```r
dir('./Normal1/sgRNA_identity/filter_after')
data_exp_sgRNA <- read.csv('./Normal1/sgRNA_identity/filter_after/cell_sgRNA_identity.csv',row.names = 1) %>% 
  rename_with(~str_remove_all(., "X"))
data_exp_sgRNA %>% dim()
meta_data <- data_exp_sgRNA %>% filter(rownames(.) == 'sgRNA_identity') %>% 
  t() %>% as.data.frame()
sgRNA <- CreateAssayObject(data_exp_sgRNA%>% filter(rownames(.) != 'sgRNA_identity'))
```

### 读取蛋白与mRNA数据

```r
dir('./Normal1/Ab-1_L4_S2401B2401_DBEC_MolsPerCell_MEX')
```

```r
count_matrix <- Read10X(data.dir = './Normal1/Ab-1_L4_S2401B2401_DBEC_MolsPerCell_MEX', gene.column = 1)
```

```r
scdata_normal_1 <- CreateSeuratObject(count_matrix[["Gene Expression"]])
adt <- CreateAssayObject(count_matrix[["Antibody Capture"]])
all.equal(colnames(count_matrix[["Gene Expression"]]), colnames(count_matrix[["Antibody Capture"]]))
scdata_normal_1[["Protein"]] <- adt
scdata_normal_1[['sgRNA']]  <- sgRNA
scdata_normal_1[['sgRNA']] %>% rownames()
Assays(scdata_normal_1)
```

```r
meta_data_all <- scdata_normal_1@meta.data %>% 
    rownames_to_column('sample_id') %>% 
    left_join(meta_data %>% rownames_to_column('sample_id'),by = 'sample_id') %>% 
    mutate(
        sgRNA_type = case_when(
            grepl('^NT\\d+$',sgRNA_identity) ~ 'NT',
            TRUE ~ sgRNA_identity %>% str_remove('-sg1|-sg2|-sg3')
        )
    ) %>% 
    column_to_rownames('sample_id')
table(!is.na(meta_data_all$sgRNA_identity))
table(meta_data_all$sgRNA_identity)
table(meta_data_all$sgRNA_type)
scdata_normal_1@meta.data <- meta_data_all
```

```r
DefaultAssay(scdata_normal_1)
```

```r
scdata_normal_1@assays$sgRNA %>% rownames()
scdata_normal_1@assays$Protein %>% rownames()
scdata_normal_1@assays$RNA %>% rownames() %>% head()
```

```r
scdata_normal_1@meta.data %>% head()
```

```r
scdata_normal_1$sgRNA_identity %>% unique()
```

```r
scdata_normal_1 <- subset(scdata_normal_1,cells = colnames(scdata_normal_1)[!is.na(scdata_normal_1$sgRNA_identity)] )
scdata_normal_1
```
